```python
Q1 = sessions["session_duration_seconds"].quantile(0.25)
Q3 = sessions["session_duration_seconds"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - (1.5 * IQR)
upper_limit = Q3 + (1.5 * IQR)

outliers = sessions[
    (sessions["session_duration_seconds"] < lower_limit) |
    (sessions["session_duration_seconds"] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

print("Potential Outlier Sessions:", len(outliers))

display(
    outliers[
        ["session_id", "user_id", "session_duration_seconds", "device"]
    ].head(10)
)
```

```python
# 1. Duplicate Event IDs
print("Duplicate Event IDs:",
      events["event_id"].duplicated().sum())

# 2. Check user IDs against Users dataset
valid_users = set(task2_users["user_id"])

invalid_users = events[
    ~events["user_id"].isin(valid_users)
]

print("Invalid User IDs:", len(invalid_users))

# 3. Check session IDs against Sessions dataset
valid_sessions = set(sessions["session_id"])

invalid_sessions = events[
    ~events["session_id"].isin(valid_sessions)
]

print("Invalid Session IDs:", len(invalid_sessions))

# 4. Check whether a session is linked to multiple users
session_user_counts = (
    events.groupby("session_id")["user_id"].nunique()
)

print("Sessions linked to multiple users:",
      (session_user_counts > 1).sum())
```

```python
# 1. Check event timestamp against session start and end

event_session_check = events.merge(
    sessions[["session_id", "session_start", "session_end"]],
    on="session_id",
    how="left",
    validate="many_to_one"
)

before_start = (
    event_session_check["event_datetime"] <
    event_session_check["session_start"]
).sum()

after_end = (
    event_session_check["event_datetime"] >
    event_session_check["session_end"]
).sum()

print("Events before session start:", before_start)
print("Events after session end:", after_end)


# 2. Check sequence number validity

print("\nMinimum Sequence Number:",
      events["sequence_number"].min())

print("Maximum Sequence Number:",
      events["sequence_number"].max())

print("Zero or Negative Sequence Numbers:",
      (events["sequence_number"] <= 0).sum())


# 3. Check repeated sequence numbers within sessions

repeated_sequences = events.duplicated(
    subset=["session_id", "sequence_number"],
    keep=False
)

print("\nEvents with repeated sequence numbers:",
      repeated_sequences.sum())


# 4. Check sequence ordering against event timestamps

ordered_events = events.sort_values(
    ["session_id", "event_datetime"]
).copy()

sequence_changes = (
    ordered_events.groupby("session_id")["sequence_number"]
    .diff()
)

print("Sequence order decreases:",
      (sequence_changes < 0).sum())
```

```python
# 1. Number of sessions containing repeated sequence numbers

repeated_sequence_sessions = (
    events.groupby("session_id")["sequence_number"]
    .apply(lambda x: x.duplicated().any())
)

print("Sessions with repeated sequence numbers:",
      repeated_sequence_sessions.sum())


# 2. Check number of events and unique sequence values per session

sequence_summary = (
    events.groupby("session_id")
    .agg(
        total_events=("event_id", "count"),
        unique_sequences=("sequence_number", "nunique"),
        min_sequence=("sequence_number", "min"),
        max_sequence=("sequence_number", "max")
    )
)

display(sequence_summary.head(10))


# 3. Inspect one session with repeated sequence numbers

repeated_session_ids = repeated_sequence_sessions[
    repeated_sequence_sessions
].index

if len(repeated_session_ids) > 0:
    sample_session = repeated_session_ids[0]

    sample_events = events[
        events["session_id"] == sample_session
    ].sort_values("event_datetime")

    display(sample_events[
        ["event_id", "session_id", "event_datetime",
         "feature_name", "event_name", "sequence_number"]
    ])


# 4. Check sequence ordering after sorting by sequence number

sequence_sorted = events.sort_values(
    ["session_id", "sequence_number", "event_datetime"]
)

sequence_time_diff = (
    sequence_sorted.groupby("session_id")["event_datetime"].diff()
)

print("Timestamp decreases after sorting by sequence:",
      (sequence_time_diff < pd.Timedelta(0)).sum())
```

```python
# Analyse repeated sequence numbers by session

repeated_seq_data = events[
    events.duplicated(
        subset=["session_id", "sequence_number"],
        keep=False
    )
].copy()

# Check repeated sequence values across different event names
sequence_event_check = (
    repeated_seq_data.groupby(
        ["session_id", "sequence_number"]
    )
    .agg(
        event_count=("event_id", "count"),
        unique_event_names=("event_name", "nunique"),
        unique_features=("feature_name", "nunique")
    )
    .reset_index()
)

print("Repeated sequence groups:",
      len(sequence_event_check))

print("Groups with multiple event names:",
      (sequence_event_check["unique_event_names"] > 1).sum())

print("Groups with multiple features:",
      (sequence_event_check["unique_features"] > 1).sum())

display(sequence_event_check.head(10))
```

```python
# Task status vs completion timestamp validation

task_timestamp_check = pd.crosstab(
    task_activity["task_status"],
    task_activity["task_completed_at"].isna()
)

task_timestamp_check.columns = [
    "Timestamp_Available",
    "Timestamp_Missing"
]

display(task_timestamp_check)
```

```python
# Convert task completion timestamp to date
task_activity["task_date"] = (
    task_activity["task_viewed_at"].dt.date
)

# Aggregate task activity by user and date
task_daily_summary = (
    task_activity.groupby(["user_id", "task_date"])
    .agg(
        actual_task_count=("activity_id", "count"),
        actual_completed_count=(
            "task_status",
            lambda x: (x == "Completed").sum()
        )
    )
    .reset_index()
)

print("Task Daily Summary Shape:",
      task_daily_summary.shape)

display(task_daily_summary.head(10))
```

```python
# 1. Task count based on viewed date
task_viewed_summary = (
    task_activity
    .groupby([
        "user_id",
        task_activity["task_viewed_at"].dt.date
    ])
    .size()
    .reset_index(name="actual_task_count")
)

task_viewed_summary.columns = [
    "user_id", "activity_date", "actual_task_count"
]


# 2. Completed task count based on completion date
completed_tasks = task_activity[
    task_activity["task_status"] == "Completed"
].copy()

completed_tasks["completion_date"] = (
    completed_tasks["task_completed_at"].dt.date
)

task_completed_summary = (
    completed_tasks
    .groupby(["user_id", "completion_date"])
    .size()
    .reset_index(name="actual_completed_count")
)

task_completed_summary.columns = [
    "user_id", "activity_date", "actual_completed_count"
]


# 3. Prepare Daily Activity dates
daily_check = daily_activity.copy()

daily_check["activity_date"] = (
    daily_check["activity_date"].dt.date
)


# 4. Merge both task summaries
daily_check = daily_check.merge(
    task_viewed_summary,
    on=["user_id", "activity_date"],
    how="left",
    validate="one_to_one"
)

daily_check = daily_check.merge(
    task_completed_summary,
    on=["user_id", "activity_date"],
    how="left",
    validate="one_to_one"
)


# 5. Fill missing task records with zero
daily_check[
    ["actual_task_count", "actual_completed_count"]
] = daily_check[
    ["actual_task_count", "actual_completed_count"]
].fillna(0)


# 6. Compare counts
daily_check["task_count_match"] = (
    daily_check["task_count"] ==
    daily_check["actual_task_count"]
)

daily_check["completed_count_match"] = (
    daily_check["completed_task_count"] ==
    daily_check["actual_completed_count"]
)


print("Total Daily Activity Records:", len(daily_check))

print("\nTask Count Matches:",
      daily_check["task_count_match"].sum())

print("Task Count Mismatches:",
      (~daily_check["task_count_match"]).sum())

print("\nCompleted Count Matches:",
      daily_check["completed_count_match"].sum())

print("Completed Count Mismatches:",
      (~daily_check["completed_count_match"]).sum())
```

```python
# Show task count mismatches
task_mismatch_sample = daily_check[
    ~daily_check["task_count_match"]
][[
    "user_id",
    "activity_date",
    "task_count",
    "actual_task_count"
]].copy()

print("Task Count Mismatch Samples:")
display(task_mismatch_sample.head(15))


# Show completed task count mismatches
completed_mismatch_sample = daily_check[
    ~daily_check["completed_count_match"]
][[
    "user_id",
    "activity_date",
    "completed_task_count",
    "actual_completed_count"
]].copy()

print("\nCompleted Task Count Mismatch Samples:")
display(completed_mismatch_sample.head(15))
```

```python
feature_usage = pd.read_excel(
    file2,
    sheet_name="07_Feature_Usage"
)

print("Shape:", feature_usage.shape)

print("\nColumns:")
print(feature_usage.columns.tolist())

print("\nFirst 5 records:")
display(feature_usage.head())

print("\nData Types:")
display(feature_usage.dtypes)

print("\nMissing Values:")
display(feature_usage.isnull().sum())

print("\nDuplicate Rows:",
      feature_usage.duplicated().sum())
```

```python
# Calculate feature-level metrics from raw Feature Events
feature_usage_check = (
    events.groupby("feature_name")
    .agg(
        calculated_unique_users=("user_id", "nunique"),
        calculated_total_events=("event_id", "count"),
        calculated_sessions=("session_id", "nunique")
    )
    .reset_index()
)

# Calculate users with at least one Success event
success_users = (
    events[events["event_status"] == "Success"]
    .groupby("feature_name")["user_id"]
    .nunique()
    .reset_index(name="calculated_success_users")
)

feature_usage_check = feature_usage_check.merge(
    success_users,
    on="feature_name",
    how="left"
)

feature_usage_check["calculated_success_users"] = (
    feature_usage_check["calculated_success_users"].fillna(0).astype(int)
)

# Compare with the supplied summary
comparison = feature_usage.merge(
    feature_usage_check,
    on="feature_name",
    how="outer",
    validate="one_to_one"
)

comparison["unique_users_match"] = (
    comparison["unique_users"] ==
    comparison["calculated_unique_users"]
)

comparison["total_events_match"] = (
    comparison["total_events"] ==
    comparison["calculated_total_events"]
)

comparison["sessions_match"] = (
    comparison["sessions_with_feature"] ==
    comparison["calculated_sessions"]
)

comparison["success_users_match"] = (
    comparison["users_with_success_event"] ==
    comparison["calculated_success_users"]
)

display(comparison)

print("\nMetric Match Summary:")
display(
    comparison[
        [
            "unique_users_match",
            "total_events_match",
            "sessions_match",
            "success_users_match"
        ]
    ].sum()
)
```

```python
# Recalculate average metrics
feature_usage_check["calculated_events_per_user"] = (
    feature_usage_check["calculated_total_events"] /
    feature_usage_check["calculated_unique_users"]
).round(2)

feature_usage_check["calculated_sessions_per_user"] = (
    feature_usage_check["calculated_sessions"] /
    feature_usage_check["calculated_unique_users"]
).round(2)

# Compare with supplied summary
average_comparison = feature_usage.merge(
    feature_usage_check[
        [
            "feature_name",
            "calculated_events_per_user",
            "calculated_sessions_per_user"
        ]
    ],
    on="feature_name",
    how="inner",
    validate="one_to_one"
)

average_comparison["events_per_user_match"] = (
    average_comparison["events_per_user"] ==
    average_comparison["calculated_events_per_user"]
)

average_comparison["sessions_per_user_match"] = (
    average_comparison["sessions_per_user"] ==
    average_comparison["calculated_sessions_per_user"]
)

display(average_comparison)

print("\nAverage Metric Match Summary:")
display(
    average_comparison[
        [
            "events_per_user_match",
            "sessions_per_user_match"
        ]
    ].sum()
)
```
